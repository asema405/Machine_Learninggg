```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Настройка визуализации
sns.set_theme(style="whitegrid")

# Загрузка данных по относительному пути
df = pd.read_csv('../data/tips.csv')
print("Данные успешно загружены!")
```

```python
print("--- 1. Размер таблицы (строки, столбцы) ---")
print(f"Количество строк: {df.shape[0]}")
print(f"Количество столбцов: {df.shape[1]}")

print("\n--- 2. Типы столбцов ---")
print(df.dtypes)

print("\n--- 3. Количество пропусков по столбцам ---")
print(df.isnull().sum())
```

```python
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

sns.histplot(df['total_bill'], kde=True, ax=axes[0], color='skyblue')
axes[0].set_title('Распределение суммы общего счета (total_bill)')
axes[0].set_xlabel('Сумма счета ($)')
axes[0].set_ylabel('Частота')

# График 2: Соотношение категорий (Количество посещений по дням недели / day)
sns.countplot(data=df, x='day', palette='pastel', ax=axes[1],
              order=['Thur', 'Fri', 'Sat', 'Sun'])
axes[1].set_title('Распределение посещений по дням недели')
axes[1].set_xlabel('День недели')
axes[1].set_ylabel('Количество заказов')

plt.tight_layout()
plt.show()
```

### 📊 Содержательные наблюдения о данных:

1. **Правосторонняя асимметрия счетов:** Распределение суммы общего счета (`total_bill`) имеет выраженный скос вправо. Большинство счетов находятся в диапазоне от $10 до $25, однако встречаются редкие крупные чеки (более $40–50).
2. **Пиковые дни посещаемости:** Наибольший поток клиентов и заказов приходится на выходные дни (суббота — `Sat` и воскресенье — `Sun`). Пятница (`Fri`) демонстрирует наименьшее число записей в датасете.
3. **Отсутствие пропусков:** Датасет является полностью полным, не содержит пропущенных значений (`null`), что позволяет сразу переходить к анализу без дополнительной предобработки.
